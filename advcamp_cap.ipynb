```python
y = df_4_oh['units']
X = df_4_oh.drop(['units'], axis = 1)

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 1027)
```

```python
from sklearn.preprocessing import RobustScaler , MinMaxScaler, StandardScaler
rs = RobustScaler()

x_train = rs.fit_transform(X_train)
x_test = rs.transform(X_test)
```

```python

from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.ensemble import RandomForestRegressor, AdaBoostRegressor
from xgboost import XGBRegressor, XGBRFRegressor
from sklearn.model_selection import GridSearchCV
import joblib

lir = LinearRegression()
lor = LogisticRegression()
rfr = RandomForestRegressor()
abr = AdaBoostRegressor()
xgbr = XGBRegressor()

xgbrfr = XGBRFRegressor(random_state=42,n_jobs=1,colsample_bynode=0.8,learning_rate=1.0,reg_lambda=0.1,subsample=0.6
)

lir.fit(x_train, y_train)
lor.fit(x_train, y_train)
rfr.fit(x_train, y_train)
abr.fit(x_train, y_train)
xgbr.fit(x_train, y_train)
xgbrfr.fit(x_train, y_train)

xgbrfr.save_model("xgbrfr_model.json")
joblib.dump(rs, "robust_scaler.pkl")

print("XGBoost 모델 저장 완료!")
print("RobustScaler 저장 완료!")
```

```python
y_pred_lir = lir.predict(x_test)
y_pred_lor = lor.predict(x_test)
y_pred_rfr = rfr.predict(x_test)
y_pred_abr = abr.predict(x_test)
y_pred_xgbr = xgbr.predict(x_test)
y_pred_xgbrfr = xgbrfr.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, accuracy_score

mae_lir = mean_absolute_error(y_test, y_pred_lir)
mae_lor = mean_absolute_error(y_test, y_pred_lor)
mae_rfr = mean_absolute_error(y_test, y_pred_rfr)
mae_abr = mean_absolute_error(y_test, y_pred_abr)
mae_xgbr = mean_absolute_error(y_test, y_pred_xgbr)
mae_xgbrfr = mean_absolute_error(y_test, y_pred_xgbrfr)

mse_xgbrfr = mean_squared_error(y_test, y_pred_xgbrfr)
mae_xgbrfr
```

```python
wape_lir = mae_lir / np.mean(y_test)
wape_lor = mae_lor / np.mean(y_test)
wape_rfr = mae_rfr / np.mean(y_test)
wape_abr = mae_abr / np.mean(y_test)
wape_xgbr = mae_xgbr / np.mean(y_test)
wape_xgbrfr = mae_xgbrfr / np.mean(y_test)

wape_percentage = wape_xgbrfr * 100

print(f"MAE: {mae_xgbrfr}")
print(f"WAPE: {wape_percentage:.2f}%")
```

```python
from xgboost import XGBRFRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import mean_absolute_error, r2_score

# 1. 모델 생성
model = XGBRFRegressor(
    random_state=42,
    n_jobs=1
)

# 2. 하이퍼파라미터 범위 설정
param_grid = {
    # 1. 트리 개수
    #'n_estimators': [100, 200, 300, 500, 700],

    # 2. 트리 최대 깊이
    #'max_depth': [3, 5, 7, 9, 12],

    # 3. 학습률 (XGBRF는 기본적으로 1.0 권장)
    'learning_rate': [0.5, 0.7, 1.0],

    # 4. 데이터 샘플링 비율
    'subsample': [0.6, 0.7, 0.8, 0.9, 1.0],

    # 5. 트리별 특성 샘플링 비율
    #'colsample_bytree': [0.6, 0.8, 1.0],

    # 6. 노드별 특성 샘플링 비율
    'colsample_bynode': [0.6, 0.8, 1.0],

    # 7. 자식 노드의 최소 가중치 합
    #'min_child_weight': [1, 2, 3, 5, 10],

    # 8. 노드 분할에 필요한 최소 손실 감소량
    #'gamma': [0, 0.1, 0.3, 0.5, 1.0],

    # 9. L1 정규화
    #'reg_alpha': [0, 0.01, 0.1, 1, 5],

    # 10. L2 정규화
    'reg_lambda': [0.1, 0.5, 1, 2, 5, 10],

    # 11. 최대 리프 노드 개수
    #'max_leaves': [0, 16, 32, 64],

    # 12. 트리 생성 알고리즘
    #'tree_method': ['hist', 'approx']
}

# 3. GridSearchCV
grid_search = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    scoring='neg_mean_absolute_error',
    cv=5,
    n_jobs=-1,
    verbose=1
)

# 4. 학습
grid_search.fit(x_train, y_train)

# 5. 최적 파라미터
print("Best Parameters:", grid_search.best_params_)
print("Best CV MAE:", -grid_search.best_score_)

# 6. 최적 모델
best_model = grid_search.best_estimator_

# 7. 테스트 데이터 평가
y_pred = best_model.predict(x_test)

print("Test MAE:", mean_absolute_error(y_test, y_pred))
print("Test R2:", r2_score(y_test, y_pred))

#Best Parameters: 'colsample_bynode': 0.8, 'learning_rate': 1.0, 'reg_lambda': 0.1, 'subsample': 0.6
```
